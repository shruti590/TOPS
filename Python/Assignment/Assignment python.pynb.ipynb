# Write a Python program to count the number of characters (character frequency) in a string.

```python
def char_frequency(text):
    freq = {}
    for char in text:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
    return freq

# Example usage
string = input("Enter a string: ")
result = char_frequency(string)

print("Character frequency:")
for char, count in result.items():
    print(f"'{char}': {count}")
```

# Write a Python program to get a single string from two given strings, separated by a space and swap the first two characters of each string. 

```python
# Program to swap first two characters of two strings and join them

def swap_and_join(str1, str2):
    # Ensure both strings have at least 2 characters
    if len(str1) < 2 or len(str2) < 2:
        return "Both strings must have at least 2 characters."
    
    # Swap first two characters
    new_str1 = str2[:2] + str1[2:]
    new_str2 = str1[:2] + str2[2:]
    
    # Join with space
    return new_str1 + " " + new_str2

# Example usage
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

result = swap_and_join(s1, s2)
print("Result:", result)
```

#  Write a Python program to count the number of strings where the string length is 2 or more and the first and last character are same from a given list of strings. 

```python
# Program to count strings with length >= 2 and same first & last character

def count_strings(words):
    count = 0
    for word in words:
        if len(word) >= 2 and word[0] == word[-1]:
            count += 1
    return count

# Example usage
string_list = ['abc', 'xyz', 'aba', '1221', 'aa', 'p']
result = count_strings(string_list)

print("Number of strings:", result)
```

# Write a Python program to find the second smallest number in a list.

```python
# # Program to find the second smallest number in a list

def second_smallest(lst):
    if len(lst) < 2:
        return "List must have at least two elements."
    
    unique_lst = list(set(lst))  # remove duplicates
    if len(unique_lst) < 2:
        return "No second smallest element (all elements are same)."
    
    unique_lst.sort()
    return unique_lst[1]

# Example usage
numbers = [5, 3, 8, 1, 9, 1, 3]
print("List:", numbers)
print("Second smallest number:", second_smallest(numbers))
```

# Write a Python program to check whether a list contains a sub list.

```python
# # Program to check if a list contains a sublist

def contains_sublist(main_list, sub_list):
    n, m = len(main_list), len(sub_list)
    if m == 0:   # empty sublist always exists
        return True
    for i in range(n - m + 1):
        if main_list[i:i+m] == sub_list:
            return True
    return False

# Example usage
main = [1, 2, 3, 4, 5, 6]
sub1 = [3, 4, 5]
sub2 = [2, 5]

print("Main List:", main)
print("Contains", sub1, "?", contains_sublist(main, sub1))  # True
print("Contains", sub2, "?", contains_sublist(main, sub2))  # False
```

# Write a Python script to concatenate following dictionaries to create a new one.

```python
# Program to concatenate multiple dictionaries into a new one

dict1 = {1: "a", 2: "b"}
dict2 = {3: "c", 4: "d"}
dict3 = {5: "e", 6: "f"}

# Method 1: Using update()
new_dict = {}
for d in (dict1, dict2, dict3):  # iterate through each dictionary
    new_dict.update(d)

print("Concatenated Dictionary:", new_dict)

# Method 2 (Python 3.5+): Using unpacking
new_dict2 = {**dict1, **dict2, **dict3}
print("Concatenated Dictionary (using **):", new_dict2)
```

#  Write a Python program to map two lists into a dictionary Sample output: Counter ({'a': 400, 'b': 400,’d’: 400, 'c': 300}). 

```python
from collections import Counter

keys = ['a', 'b', 'c', 'd', 'b']
values = [100, 200, 300, 400, 200]

# Map lists into a dictionary (with Counter to handle duplicates)
result = Counter(dict(zip(keys, values)))  # simple mapping
print("Mapped:", result)

# If you want to sum values of duplicate keys
result = Counter()
for k, v in zip(keys, values):
    result[k] += v

print("Combined:", result)
```

# Write a Python function to calculate the factorial of a number (a nonnegative integer) 

```python
def factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers"
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact

# Example
print(factorial(5))  
```

# Write a Python function to check whether a number is perfect or not. 

```python
def is_perfect(n):
    if n <= 0:
        return False
    divisors_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisors_sum += i
    return divisors_sum == n

# Examples
print(is_perfect(6))   # True
print(is_perfect(28))  # True
#print(is_perfect(12))  # False
```

#  Write a Python program to read first n lines of a file. 

```python
def read_first_n_lines(filename, n):
    with open(filename, "r") as file:
        for i in range(n):
            line = file.readline()
            if not line:   # stop if file has fewer than n lines
                break
            print(line.strip())

# Example usage
read_first_n_lines("example.txt", 3)
```

# Write a Python program to read last n lines of a file.

```python
def read_last_n_lines(filename, n):
    with open(filename, "r") as file:
        lines = file.readlines()   # read all lines into a list
        for line in lines[-n:]:    # slice last n lines
            print(line.strip())

# Example usage
read_last_n_lines("example.txt", 3)
```

#  Write a Python program to read a file line by line and store it into a list.

```python
def file_to_list(filename):
    with open(filename, "r") as file:
        lines = file.readlines()        # read all lines
        lines = [line.strip() for line in lines]  # remove \n
    return lines

# Example usage
lines_list = file_to_list("example.txt")
print(lines_list)
```

#  Write a python program to find the longest words.

```python
# Program to find the longest words in a file

with open("Assignment python.pynb.ipynb", "r") as f:
    words = f.read().split()

max_len = max(len(word) for word in words)
longest_words = [word for word in words if len(word) == max_len]

print("Longest word(s):", longest_words)
```

#  Write a Python program to write a list to a file. 

```python
# Program to write a list to a file

my_list = ["Apple", "Banana", "Cherry", "Mango"]

with open("output.txt", "w") as f:
    for item in my_list:
        f.write(item + "\n")

print("List written to file successfully!")
```

# Write python program that user to enter only odd numbers, else will raise an exception. 

```python
try:
    num = int(input("Enter an odd number: "))
    if num % 2 == 0:
        raise ValueError(" Only odd numbers are allowed!")
    print(" You entered an odd number:", num)
except ValueError as e:
    print("Error:", e)
```
