```python
import numpy as np


doctors = np.array(["Dr. Hathi", "Dr. Mehta", "Dr. Patel"])
slots = np.array (["10am", "11am", "12pm", "1pm", "2pm"])

slot_counts = np.zeros((len(doctors), len(slots)), dtype=int)

appointments = {doc: {s: [] for s in slots} for doc in doctors}


def book_appointment():
    name = input("Enter patient name: ")
    age = input("Enter age: ")
    mobile = input("Enter mobile number: ")

    print("\nAvailable doctors:")
    for i, d in enumerate(doctors, 1):
        print(f"{i}. {d}")
    d_choice = int(input("Select doctor (1-3): ")) - 1
    doctor = doctors[d_choice]

    print("\nAvailable slots:")
    for i, s in enumerate(slots, 1):
        print(f"{i}. {s}")

    slot_input = input("Select slot (number 1-5 or time like '12pm'): ").strip()

    if slot_input.isdigit():
        s_choice = int(slot_input) - 1
        if s_choice < 0 or s_choice >= len(slots):
            print("Invalid slot number!")
            return
        slot = slots[s_choice]
    else:
        if slot_input not in slots:
            print("Invalid slot name!")
            return
        slot = slot_input
        s_choice = slots.index(slot)

    if slot_counts[d_choice, s_choice] < 3:
        appointments[doctor][slot].append({
            "name": name,
            "age": age,
            "mobile": mobile
        })
        slot_counts[d_choice, s_choice] += 1
        print(f"\n Appointment booked with {doctor} at {slot} for {name}.")
    else:
        print("\n Slot full! Please choose another slot.")


def view_or_cancel():
    mobile = input("Enter your mobile number: ")
    found = False

    for d_idx, doctor in enumerate(doctors):
        for s_idx, slot in enumerate(slots):
            for appt in appointments[doctor][slot]:
                if appt["mobile"] == mobile:
                    found = True
                    print(f"\n Appointment found:\nDoctor: {doctor}, Slot: {slot}, Name: {appt['name']}, Age: {appt['age']}")
                    choice = input("Do you want to cancel? (y/n): ")
                    if choice.lower() == "y":
                        appointments[doctor][slot].remove(appt)
                        slot_counts[d_idx, s_idx] -= 1
                        print(" Appointment cancelled.")
                    return
    if not found:
        print(" No appointment found for this number.")


def show_all():
    print("\n Current Appointments:")
    for doctor in doctors:
        for slot in slots:
            if appointments[doctor][slot]:
                print(f"\nDoctor: {doctor}, Slot: {slot}")
                for appt in appointments[doctor][slot]:
                    print(f" - {appt['name']} ({appt['age']} yrs, {appt['mobile']})")

    print("\n Slot usage (per doctor-slot):")
    print(slot_counts)



while True:
    print("\n--- Appointment System ---")
    print("1. Book Appointment")
    print("2. View/Cancel Appointment")
    print("3. Show All Appointments")
    print("4. Exit")

    choice = input("Enter choice: ")
    if choice == "1":
        book_appointment()
    elif choice == "2":
        view_or_cancel()
    elif choice == "3":
        show_all()
    elif choice == "4":
        print("Goodbye!")
        break
    else:
        print("Invalid choice!")
```

```python
import numpy as np

students = {}
student_ids = np.array([], dtype=int)
next_id = 1

def new_admission():
    global next_id, student_ids

    name = input("Enter student name: ").strip()
    age = int(input("Enter age: "))
    if age < 5 or age > 18:
        print("Invalid age! Must be between 5 and 18.")
        return

    class_no = int(input("Enter class (1-12): "))
    if class_no < 1 or class_no > 12:
        print(" Invalid class! Must be between 1 and 12.")
        return

    mobile = input("Enter guardian's mobile number: ").strip()
    if not (mobile.isdigit() and len(mobile) == 10):
        print(" Invalid mobile number! Must be 10 digits.")
        return


    sid = next_id
    next_id += 1
    student_ids = np.append(student_ids, sid)

    students[sid] = {
        "name": name,
        "age": age,
        "class": class_no,
        "mobile": mobile
    }

    print(f"\n Admission successful! Student ID = {sid}")


def view_student():
    sid = int(input("Enter student ID: "))
    if sid in students:
        s = students[sid]
        print(f"\n Student Details [ID {sid}]:")
        print(f"Name: {s['name']}")
        print(f"Age: {s['age']}")
        print(f"Class: {s['class']}")
        print(f"Guardian Mobile: {s['mobile']}")
    else:
        print("Student ID not found.")


def update_student():
    sid = int(input("Enter student ID to update: "))
    if sid not in students:
        print(" Student ID not found.")
        return

    print("\nWhat do you want to update?")
    print("1. Mobile Number")
    print("2. Class")
    choice = input("Enter choice: ")

    if choice == "1":
        mobile = input("Enter new mobile number: ").strip()
        if not (mobile.isdigit() and len(mobile) == 10):
            print(" Invalid mobile number! Must be 10 digits.")
            return
        students[sid]["mobile"] = mobile
        print(" Mobile number updated.")
    elif choice == "2":
        class_no = int(input("Enter new class (1-12): "))
        if class_no < 1 or class_no > 12:
            print(" Invalid class! Must be between 1 and 12.")
            return
        students[sid]["class"] = class_no
        print(" Class updated.")
    else:
        print(" Invalid choice.")


def remove_student():
    global student_ids
    sid = int(input("Enter student ID to remove: "))
    if sid in students:
        del students[sid]
        student_ids = student_ids[student_ids != sid]  # remove from numpy array
        print(f" Student ID {sid} removed.")
    else:
        print("Student ID not found.")


def show_all():
    if not students:
        print("\n No student records available.")
        return

    print("\n All Students:")
    for sid in sorted(students.keys()):
        s = students[sid]
        print(f"ID {sid}: {s['name']} | Age: {s['age']} | Class: {s['class']} | Mobile: {s['mobile']}")

    print("\n All Student IDs (NumPy):", student_ids)


while True:
    print("\n--- Student Admission System ---")
    print("1. New Admission")
    print("2. View Student Details")
    print("3. Update Student Info")
    print("4. Remove Student Record")
    print("5. Show All Students")
    print("6. Exit")

    choice = input("Enter choice: ")
    if choice == "1":
        new_admission()
    elif choice == "2":
        view_student()
    elif choice == "3":
        update_student()
    elif choice == "4":
        remove_student()
    elif choice == "5":
        show_all()
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print("Invalid choice!")
```

```python
import numpy as np


routes = {
    "Mumbai to Pune": 500,
    "Delhi to Jaipur": 600
}


tickets = {}
ticket_ids = np.array([], dtype=int)
next_ticket_id = 1

seats = {route: np.zeros(40, dtype=int) for route in routes}


def show_routes():
    print("\n🚌 Available Routes:")
    for i, (route, price) in enumerate(routes.items(), 1):
        print(f"{i}. {route} - ₹{price}")


def book_ticket():
    global next_ticket_id, ticket_ids

    name = input("Enter passenger name: ").strip()
    age = int(input("Enter age: "))
    mobile = input("Enter mobile number: ").strip()

    if not (mobile.isdigit() and len(mobile) == 10):
        print(" Invalid mobile number! Must be 10 digits.")
        return

    show_routes()
    route_input = input("Enter route (exact name or number): ").strip()

    if route_input.isdigit():
        r_idx = int(route_input) - 1
        if r_idx < 0 or r_idx >= len(routes):
            print(" Invalid route number!")
            return
        route = list(routes.keys())[r_idx]
    else:
        if route_input not in routes:
            print(" Invalid route name!")
            return
        route = route_input


    available_seats = np.where(seats[route] == 0)[0]
    if len(available_seats) == 0:
        print(" No seats available for this route.")
        return

    seat_no = available_seats[0] + 1
    seats[route][seat_no - 1] = 1

    ticket_id = next_ticket_id
    next_ticket_id += 1
    ticket_ids = np.append(ticket_ids, ticket_id)

    tickets[ticket_id] = {
        "name": name,
        "age": age,
        "mobile": mobile,
        "route": route,
        "seat": seat_no,
        "price": routes[route]
    }

    print(f"\n Ticket booked successfully!")
    print(f"Ticket ID: {ticket_id}, Seat: {seat_no}, Route: {route}, Price: ₹{routes[route]}")


def view_ticket():
    tid = int(input("Enter Ticket ID: "))
    if tid in tickets:
        t = tickets[tid]
        print(f"\n Ticket Details [ID {tid}]:")
        print(f"Name: {t['name']}")
        print(f"Age: {t['age']}")
        print(f"Mobile: {t['mobile']}")
        print(f"Route: {t['route']}")
        print(f"Seat: {t['seat']}")
        print(f"Price: ₹{t['price']}")
    else:
        print(" Ticket ID not found.")


def cancel_ticket():
    global ticket_ids
    tid = int(input("Enter Ticket ID to cancel: "))
    if tid in tickets:
        route = tickets[tid]["route"]
        seat = tickets[tid]["seat"]
        seats[route][seat - 1] = 0
        del tickets[tid]
        ticket_ids = ticket_ids[ticket_ids != tid]
        print(f" Ticket ID {tid} cancelled.")
    else:
        print(" Ticket ID not found.")


def show_all_tickets():
    if not tickets:
        print("\n📋 No tickets booked.")
        return
    print("\n📋 All Tickets:")
    for tid, t in tickets.items():
        print(f"ID {tid}: {t['name']} | Age: {t['age']} | Route: {t['route']} | Seat: {t['seat']} | Price: ₹{t['price']}")


while True:
    print("\n--- Bus Ticket Booking System ---")
    print("1. Show Available Routes")
    print("2. Book Ticket")
    print("3. View Ticket")
    print("4. Cancel Ticket")
    print("5. Show All Tickets")
    print("6. Exit")

    choice = input("Enter choice: ")
    if choice == "1":
        show_routes()
    elif choice == "2":
        book_ticket()
    elif choice == "3":
        view_ticket()
    elif choice == "4":
        cancel_ticket()
    elif choice == "5":
        show_all_tickets()
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print(" Invalid choice!")
```
