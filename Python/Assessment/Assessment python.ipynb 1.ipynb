```python
class ClinicAppointment:
    def __init__(self):
        self.doctors = ['Dr. Smith', 'Dr. Johnson', 'Dr. Williams']
        self.time_slots = ['10am', '11am', '12pm', '2pm', '3pm']
        self.appointments = {}

    def book_appointment(self):
        name = input("Enter patient name: ")
        age = input("Enter patient age: ")
        mobile = input("Enter mobile number: ")
        print("Available doctors: ", self.doctors)
        doctor = input("Enter preferred doctor: ")
        print("Available time slots: ", self.time_slots)
        time_slot = input("Enter preferred time slot: ")

        if doctor not in self.appointments:
            self.appointments[doctor] = {}
        if time_slot not in self.appointments[doctor]:
            self.appointments[doctor][time_slot] = []

        if len(self.appointments[doctor].get(time_slot, [])) < 3:
            self.appointments[doctor][time_slot].append({'name': name, 'age': age, 'mobile': mobile})
            print("Appointment booked successfully!")
        else:
            print("Time slot not available.")

    def view_cancel_appointment(self):
        mobile = input("Enter mobile number: ")
        for doctor, time_slots in self.appointments.items():
            for time_slot, appointments in time_slots.items():
                for appointment in appointments:
                    if appointment['mobile'] == mobile:
                        print(f"Doctor: {doctor}, Time Slot: {time_slot}, Name: {appointment['name']}")
                        action = input("Do you want to cancel? (yes/no): ")
                        if action.lower() == 'yes':
                            self.appointments[doctor][time_slot].remove(appointment)
                            print("Appointment cancelled.")
                        return
        print("No appointment found.")

    def run(self):
        while True:
            print("1. Book Appointment")
            print("2. View/Cancel Appointment")
            print("3. Exit")
            choice = input("Enter your choice: ")
            if choice == '1':
                self.book_appointment()
            elif choice == '2':
                self.view_cancel_appointment()
            elif choice == '3':
                break
            else:
                print("Invalid choice.")

if __name__ == "__main__":
    clinic = ClinicAppointment()
    clinic.run()
```

```python
class SchoolManagement:
    def __init__(self):
        self.students = {}
        self.student_id = 1

    def new_admission(self):
        name = input("Enter student name: ")
        while True:
            try:
                age = int(input("Enter student age: "))
                if 5 <= age <= 18:
                    break
                else:
                    print("Age must be between 5 and 18.")
            except ValueError:
                print("Invalid age. Please enter a number.")
        while True:
            try:
                class_ = int(input("Enter class (1-12): "))
                if 1 <= class_ <= 12:
                    break
                else:
                    print("Class must be between 1 and 12.")
            except ValueError:
                print("Invalid class. Please enter a number.")
        while True:
            mobile = input("Enter guardian's mobile number: ")
            if len(mobile) == 10 and mobile.isdigit():
                break
            else:
                print("Invalid mobile number. Please enter a 10-digit number.")
        student_id = self.student_id
        self.students[student_id] = {'name': name, 'age': age, 'class': class_, 'mobile': mobile}
        self.student_id += 1
        print(f"Student admitted with ID: {student_id}")

    def view_student_details(self):
        student_id = int(input("Enter student ID: "))
        if student_id in self.students:
            student = self.students[student_id]
            print(f"Name: {student['name']}")
            print(f"Age: {student['age']}")
            print(f"Class: {student['class']}")
            print(f"Mobile: {student['mobile']}")
        else:
            print("Student not found.")

    def update_student_info(self):
        student_id = int(input("Enter student ID: "))
        if student_id in self.students:
            print("1. Update mobile number")
            print("2. Update class")
            choice = input("Enter your choice: ")
            if choice == '1':
                while True:
                    mobile = input("Enter new mobile number: ")
                    if len(mobile) == 10 and mobile.isdigit():
                        self.students[student_id]['mobile'] = mobile
                        break
                    else:
                        print("Invalid mobile number. Please enter a 10-digit number.")
            elif choice == '2':
                while True:
                    try:
                        class_ = int(input("Enter new class (1-12): "))
                        if 1 <= class_ <= 12:
                            self.students[student_id]['class'] = class_
                            break
                        else:
                            print("Class must be between 1 and 12.")
                    except ValueError:
                        print("Invalid class. Please enter a number.")
            else:
                print("Invalid choice.")
        else:
            print("Student not found.")

    def remove_student_record(self):
        student_id = int(input("Enter student ID: "))
        if student_id in self.students:
            del self.students[student_id]
            print("Student record removed.")
        else:
            print("Student not found.")

    def run(self):
        while True:
            print("1. New Admission")
            print("2. View Student Details")
            print("3. Update Student Info")
            print("4. Remove Student Record")
            print("5. Exit")
            choice = input("Enter your choice: ")
            if choice == '1':
                self.new_admission()
            elif choice == '2':
                self.view_student_details()
            elif choice == '3':
                self.update_student_info()
            elif choice == '4':
                self.remove_student_record()
            elif choice == '5':
                break
            else:
                print("Invalid choice.")

if __name__ == "__main__":
    school = SchoolManagement()
    school.run()
```

```python
import uuid

class BusReservation:
    def __init__(self):
        self.routes = {
            "Mumbai to Pune": 500,
            "Delhi to Jaipur": 600,
            "Bangalore to Chennai": 700,
            "Hyderabad to Kolkata": 800
        }
        self.tickets = {}
        self.seats = {route: 40 for route in self.routes}

    def show_available_routes(self):
        for route, price in self.routes.items():
            print(f"{route} - ₹{price}")

    def book_ticket(self):
        print("Available routes:")
        self.show_available_routes()
        route = input("Enter route: ")
        if route in self.routes:
            if self.seats[route] > 0:
                name = input("Enter passenger name: ")
                age = input("Enter passenger age: ")
                mobile = input("Enter passenger mobile: ")
                ticket_id = str(uuid.uuid4())[:8]
                seat_number = 41 - self.seats[route]
                self.tickets[ticket_id] = {'name': name, 'age': age, 'mobile': mobile, 'route': route, 'seat_number': seat_number}
                self.seats[route] -= 1
                print(f"Ticket booked successfully! Ticket ID: {ticket_id}, Seat Number: {seat_number}")
            else:
                print("No seats available for this route.")
        else:
            print("Route not available.")

    def view_ticket(self):
        ticket_id = input("Enter ticket ID: ")
        if ticket_id in self.tickets:
            ticket = self.tickets[ticket_id]
            print(f"Name: {ticket['name']}")
            print(f"Age: {ticket['age']}")
            print(f"Mobile: {ticket['mobile']}")
            print(f"Route: {ticket['route']}")
            print(f"Seat Number: {ticket['seat_number']}")
        else:
            print("Ticket not found.")

    def cancel_ticket(self):
        ticket_id = input("Enter ticket ID: ")
        if ticket_id in self.tickets:
            route = self.tickets[ticket_id]['route']
            del self.tickets[ticket_id]
            self.seats[route] += 1
            print("Ticket cancelled successfully!")
        else:
            print("Ticket not found.")

    def run(self):
        while True:
            print("1. Show Available Routes")
            print("2. Book Ticket")
            print("3. View Ticket")
            print("4. Cancel Ticket")
            print("5. Exit")
            choice = input("Enter your choice: ")
            if choice == '1':
                self.show_available_routes()
            elif choice == '2':
                self.book_ticket()
            elif choice == '3':
                self.view_ticket()
            elif choice == '4':
                self.cancel_ticket()
            elif choice == '5':
                break
            else:
                print("Invalid choice.")

if __name__ == "__main__":
    bus = BusReservation()
    bus.run()
```
