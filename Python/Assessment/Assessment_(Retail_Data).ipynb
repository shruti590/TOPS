```python
# Import libraries
import pandas as pd

# Load the dataset
df = pd.read_csv("Retail Data.csv")

# 1. View the structure of the dataset (columns, types, missing values).

# View first few rows
print("First 5 rows of the dataset:")
print(df.head(), "\n")

# Check column names and data types
print("Column Names and Data Types:")
print(df.dtypes, "\n")

# Check for missing values
print("Missing Values in Each Column:")
print(df.isnull().sum(), "\n")

# Display basic info summary
print("Dataset Info Summary:")
print(df.info())

# Optional — check total rows and columns
print(f"\n The dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
# 2. What is the shape (rows, columns) of the dataset?
print("===== Shape of the Dataset =====")
print(f"The dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
# 3. Are there any duplicate records?
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# 4. Are there any missing or corrupted entries in Ship Date, Order Date, or numeric columns?
# Check for missing or corrupted entries in Ship Date, Order Date, or numeric columns

# Check missing values in date columns
print("Missing values in Order Date and Ship Date:")
print(df[['Order Date', 'Ship Date']].isnull().sum(), "\n")

# Try converting date columns to datetime (to detect corrupted entries)
for col in ['Order Date', 'Ship Date']:
    try:
        pd.to_datetime(df[col])
        print(f"{col} has valid date format.")
    except:
        print(f"{col} has invalid or corrupted entries.")

# Check missing values in numeric columns
numeric_cols = ['Cost Price', 'Retail Price', 'Profit Margin', 'Order Quantity',
                'Sub Total', 'Discount %', 'Discount $', 'Order Total', 'Shipping Cost', 'Total']

print("\nMissing values in numeric columns:")
print(df[numeric_cols].isnull().sum())
```

```python
# 5. Convert Order Date and Ship Date to datetime.
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

print("Conversion completed successfully!")
print(df[['Order Date', 'Ship Date']].head())
```

```python
# 6. Check for future or inconsistent shipping dates.
# Define today's date (you can use pd.Timestamp.now() for current date)
today = pd.Timestamp("2025-11-03")

# Future shipping dates (Ship Date after today)
future_ship = df[df['Ship Date'] > today]

# Inconsistent shipping dates (Ship Date before Order Date)
inconsistent_ship = df[df['Ship Date'] < df['Order Date']]

print("Number of future shipping dates:", len(future_ship))
print("Number of inconsistent shipping dates:", len(inconsistent_ship))
```

```python
# 7. Convert price columns to numeric (remove $ and commas).
price_cols = ['Cost Price', 'Retail Price', 'Profit Margin', 'Sub Total',
              'Discount %', 'Discount $', 'Order Total', 'Shipping Cost', 'Total']

for col in price_cols:
    df[col] = df[col].replace('[\$,%,]', '', regex=True).astype(float)

print("Price columns converted to numeric successfully!")
print(df[price_cols].head())
```

```python
# 8. What are the unique values in Customer Type and Order Priority?
print("Unique values in Customer Type:")
print(df['Customer Type'].unique(), "\n")

print("Unique values in Order Priority:")
print(df['Order Priority'].unique())
```

```python
# 9. What are the most common shipping modes?
print("Most common shipping modes:")
print(df['Ship Mode'].value_counts())
```

```python
# 10. Which cities have the highest number of orders?
print("Cities with the highest number of orders:")
print(df['City'].value_counts().head(10))
```

```python
# 11. What’s the range of order quantities and prices?
print("Range of Order Quantity:")
print("Min:", df['Order Quantity'].min(), " | Max:", df['Order Quantity'].max(), "\n")

price_cols = ['Cost Price', 'Retail Price', 'Order Total', 'Shipping Cost', 'Total']
print("Range of Price Columns:")
for col in price_cols:
    print(f"{col}: Min = {df[col].min()} | Max = {df[col].max()}")
```

```python
# 12. Create a new column for shipping duration.
df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days

print("New column 'Shipping Duration' created successfully!")
print(df[['Order Date', 'Ship Date', 'Shipping Duration']].head())
```

```python
# 13. Are there any orders with zero or negative total or quantity?
invalid_orders = df[(df['Order Quantity'] <= 0) | (df['Total'] <= 0)]

print("Number of orders with zero or negative total/quantity:", len(invalid_orders))
print(invalid_orders[['Order No', 'Order Quantity', 'Total']].head())
```

```python
# 14. Are all discount percentages matching discount dollar amounts?
# Calculate expected discount $ based on Sub Total and Discount %
df['Expected Discount $'] = (df['Sub Total'] * df['Discount %']) / 100

# Compare with actual Discount $
mismatch = df[abs(df['Expected Discount $'] - df['Discount $']) > 1]  # allow small rounding difference

print("Number of mismatched discount records:", len(mismatch))
print(mismatch[['Order No', 'Sub Total', 'Discount %', 'Discount $', 'Expected Discount $']].head())
```

```python
# 15. Check for mismatches in total calculation.
# Recalculate expected total: (Sub Total - Discount $) + Shipping Cost
df['Expected Total'] = (df['Sub Total'] - df['Discount $']) + df['Shipping Cost']

# Find mismatches between expected and actual total
mismatch_total = df[abs(df['Expected Total'] - df['Total']) > 1]  # allow small rounding diff

print("Number of mismatched total calculations:", len(mismatch_total))
# print(mismatch_total[['Order No', 'Sub Total', 'Discount $', 'Shipping Cost', 'Total', 'Expected Total']].head())
```

```python
# 6. How many days does it usually take to ship an order?
import pandas as pd
# Convert 'Order Date' and 'Ship Date' to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce', dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce', dayfirst=True)

# Calculate shipping duration in days
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

# Calculate average shipping time
average_shipping_days = df['Shipping Days'].mean()

print("Average Shipping Time (in days):", round(average_shipping_days, 2))
```

```python
# 7. Do longer shipping times impact profit margins?
import pandas as pd
# Check correlation between shipping days and profit margin
correlation = df['Shipping Days'].corr(df['Profit Margin'])

print("Correlation between Shipping Time and Profit Margin:", round(correlation, 2))
```

```python
# 15. . Are certain containers (e.g., Small Box, Wrap Bag) more profitable?s
import pandas as pd
import re
# Helper: extract the first numeric (int/float) found in the string
def extract_first_number(s):
    if pd.isna(s):
        return None
    s = str(s)
    # find all numeric patterns (allow decimal and negative)
    matches = re.findall(r'-?\d+\.\d+|-?\d+', s.replace(',', ''))
    if not matches:
        return None
    return float(matches[0])
# Apply extraction to create a clean numeric column
df['Profit_Margin_clean'] = df['Profit Margin'].apply(extract_first_number)

# Show how many missing after cleaning
print("\nCount missing after cleaning:", df['Profit_Margin_clean'].isna().sum())

# Now compute average profit margin by product container
# (replace 'Product Container' with the exact column name if different)
avg_by_container = df.groupby('Product Container')['Profit_Margin_clean'].mean().sort_values(ascending=False)

print("\nAverage Profit Margin by Product Container:")
print(avg_by_container)

print("\nMost profitable product container:", avg_by_container.idxmax())
```
